```python
# load the data
import numpy as np
import matplotlib.pyplot as plt

# load the data
ecg_raw = np.loadtxt('data/ECGu.txt')
print(ecg_raw.shape)
print(ecg_raw)
```

```python
# plot the data
plt.plot(ecg_raw)
plt.title('ECG signal')
plt.xlabel('samples')
plt.ylabel('amplitude')
plt.grid()
plt.show()
```
