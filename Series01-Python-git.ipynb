```python
# load the data
import numpy as np
import matplotlib.pyplot as plt

# load the data
ecg_raw = np.loadtxt('data/ECGu.txt')
print(ecg_raw.shape)
print(ecg_raw)
```

```python
# plot the data
plt.plot(ecg_raw)
plt.title('ECG signal')
plt.xlabel('samples')
plt.ylabel('amplitude')
plt.grid()
plt.show() 
```

```python
# create the t time vector in seconds
from matplotlib.pyplot import plot


sampling_frequency = 1000  # in Hz
T = 1 /sampling_frequency # sampling period in seconds => temps qui s'écoule entre 2 mesures
N = len(ecg_raw) # number of samples
start = 0 # start time in seconds
stop = (N-1)*T # stop time in seconds => parce que en python on commence à 0
t= np.linspace(start, stop, N) # time vector in seconds

plot(t, ecg_raw)
plt.title('ECG signal')
plt.xlabel('time (s)')
plt.ylabel('amplitude')
plt.grid()
plt.show()

```

```python
# rescale the ecg signal to mV knowing that 1 unit = 1024 microvolts
ecg = ecg_raw * 1024 * 1e-6 # in mV

#plot the ecg as a function of time
plt.plot(t, ecg)
# plt.title('ECG signal')
plt.xlabel('time (s)')
plt.ylabel('amplitude (mV)')
plt.grid()
plt.show()
```
